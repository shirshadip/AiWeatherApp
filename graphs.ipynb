```python
import pandas as pd
import matplotlib.pyplot as plt

def plot_weather_graph(csv_file, column, title, ylabel):
    """
    Plots a weather graph from a CSV file.

    Parameters:
        csv_file (str): Path to the CSV file.
        column (str): Column name to plot.
        title (str): Graph title.
        ylabel (str): Y-axis label.
    """

    # Read CSV
    df = pd.read_csv(csv_file)

    # Convert date column to datetime
    df["date"] = pd.to_datetime(df["date"])

    # Create graph
    plt.figure(figsize=(12, 5))
    plt.plot(df["date"], df[column], marker="o")

    plt.title(title)
    plt.xlabel("Date")
    plt.ylabel(ylabel)
    plt.xticks(rotation=45)
    plt.grid(True)

    plt.tight_layout()
    plt.show()
```

```python
plot_weather_graph(
    "weather_last_30_days.csv",
    "temp_max",
    "Maximum Temperature (30 Days)",
    "Temperature (°C)"
)
```

```python
plot_weather_graph(
    "weather_last_30_days.csv",
    "temp_min",
    "Minimum Temperature (30 Days)",
    "Temperature (°C)"
)
```

```python
plot_weather_graph(
    "weather_last_30_days.csv",
    "rain_mm",
    "Rainfall (30 Days)",
    "Rain (mm)"
)
```

```python
def rain_graph_last_30days(csv_file , column , title , ylabel):
    df = pd.read_csv(csv_file)
    
    # Convert date column to datetime
    df["date"] = pd.to_datetime(df["date"])

    # Create graph
    plt.figure(figsize=(12, 5))
    plt.bar(df["date"], df[column])

    plt.title(title)
    plt.xlabel("Date")
    plt.ylabel(ylabel)
    plt.xticks(rotation=45)
    plt.grid(True)

    plt.tight_layout()
    plt.show()
```

```python
rain_graph_last_30days("weather_last_30_days.csv","rain_mm", "rain" ," rain(mm)")
```

```python
WMO_WEATHER_CODES = {
    0: "Sunny",

    1: "Cloudy",
    2: "Cloudy",
    3: "Cloudy",

    45: "Fog",
    48: "Fog",

    51: "Drizzle",
    53: "Drizzle",
    55: "Drizzle",
    56: "Drizzle",
    57: "Drizzle",

    61: "Rainy",
    63: "Rainy",
    65: "Rainy",
    66: "Rainy",
    67: "Rainy",

    71: "Snowy",
    73: "Snowy",
    75: "Snowy",
    77: "Snowy",
    85: "Snowy",
    86: "Snowy",

    80: "Rain Showers",
    81: "Rain Showers",
    82: "Rain Showers",

    95: "Thunderstorm",
    96: "Thunderstorm",
    99: "Thunderstorm"
}
df = pd.read_csv("weather_daily_all_report.csv")
df["weather_condition"] = (
    df["weather_code"]
    .map(WMO_WEATHER_CODES)
    .fillna("Unknown")
)
weather_counts = df["weather_condition"].value_counts()

print(weather_counts)
```

```python
weather_counts.plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(7, 7),
    ylabel="",
    title="Weather Conditions (Last 30 Days)"

)
```
